## ***TRANSACTION PARSER***

```python
import pandas as pd, numpy as np
from datetime import datetime, date as dt
```

```python
# To drop the duplicates from the original dataset
df.drop_duplicates(inplace=True)
```

```python
# Parsing all the four types of dates in the Date column and storing it in a different data frame named 'date'
df['date'] = pd.to_datetime(df['Date'], errors='coerce', dayfirst=True, format='mixed')
```

```python
# To record time from 'date' column in which a transaction took place
df['hour'] = df['Time'].str[:2].astype(int)

# To record months from 'date' column in which a transaction took place
df['month'] = df['date'].dt.month

# To record specific days of the from 'date' column when a transaction took place
df['day_of_week'] = df['date'].dt.day_name()

# To record specific weeks of the calendar from 'date' column
df['week_number'] = df['date'].dt.isocalendar().week
```

```python
# To check whether all the dates have been parsed
df['date'].count()
```

```python
# To remove timestamp from 'date'
df['date'] = df['date'].dt.strftime('%Y-%m-%d')
```

```python
# To remove special characters like ',', '₹' and 'Rs.' from 'Amount' column. This is essential so as to convert Amount into numerical data.
df['Amount'] = df['Amount'].str.replace('₹', '').str.replace('Rs.', '').str.replace(',', '').str.strip()
```

```python
# Converting Amount into numerical data
df['Amount'] = pd.to_numeric(df['Amount'], errors='coerce')
```

```python
# Checking whether amount is stored as a numerical value that is a float or int type
df['Amount']
```

```python
# Replacing records 'CR' and 'DR' with 'credit' and 'debit' respectively under 'Type' column
df['Type'] = df['Type'].str.replace('DR','debit').str.replace('CR','credit').str.strip()
```

```python
# Standardising format so that records like 'Credit' and 'Debit' are stored as 'credit' and 'debit' respectively under 'Type' column
df['Type'] = df['Type'].str.lower()
```

```python
# Checking whether the previous codes fro 'Type' column ran successfully
df['Type']
```

## ***ANOMALY DETECTION***

```python
category_mean = df.groupby('category')['Amount'].transform('mean')
category_std = df.groupby('category')['Amount'].transform('std')
df['z_score'] = (df['Amount'] - category_mean) / category_std
anomalies = df[df['z_score'] > 2] # transactions 2+ stddev above category mean
```

```python
anomalies
```

```python
# Number of Anomalies
len(anomalies)
```

```python
top_2_anomalies = anomalies.sort_values(by='z_score', ascending=False).head(2)
display(top_2_anomalies[['date', 'vendor_clean', 'Amount', 'z_score']])
```

## *SPENDING ARCHETYPE SYSTEM*

```python
# Total of debit transactions for each category measured by amount debited from the account
category_totals = df[df['Type'] == 'debit'].groupby('category')['Amount'].sum()
```

```python
# Calculating total amount debited for each category based on archetype guidelines
food_debits = category_totals['Food Delivery'] + category_totals['Restaurants'] + category_totals['Cafe']
qcom_debits = category_totals['Quick Commerce']
ecom_debits = category_totals['E-commerce']
inv_debits = category_totals['Investments']
trans_debits = category_totals['Transport']
```

```python
# Calculating amount spent on each category as a percentage of total debit transactions based on archetype guidelines
food_pct = (food_debits / total_debits) * 100
qcom_pct = (qcom_debits / total_debits) * 100
ecom_pct = (ecom_debits / total_debits) * 100
inv_pct = (inv_debits / total_debits) * 100
trans_pct = (trans_debits / total_debits) * 100
```

```python
# Computing number of active subscriptions
active_sub_suppliers = df[df['category'] == 'Subscriptions']['vendor_clean'].nunique()
```

```python
# Categorising Rahul based on the following archetypes

archetypes = []
if food_pct > 25:
 archetypes.append(f"THE FOODIE ({food_pct:.1f}% on food)")
if qcom_pct > 15:
 archetypes.append(f"THE QUICK COMMERCE JUNKIE ({qcom_pct:.1f}% on Q-com)")
if ecom_pct > 15:
 archetypes.append(f"THE SHOPAHOLIC ({ecom_pct:.1f}% on e-commerce)")
if inv_pct > 15:
  archetypes.append(f"THE INVESTOR ({inv_pct:.1f}% on SIPs)")
if late_night_food > 50:
 archetypes.append(f"THE LATE-NIGHT SNACKER ({late_night_food:.0f}% food after 9 PM)")
if trans_pct > 10:
 archetypes.append(f"THE CAB COMMUTER ({trans_pct:.1f}% on cab rides)")
if active_sub_suppliers >= 5:
 archetypes.append(f"THE SUBSCRIPTION LOVER ({active_sub_suppliers} active subs)")
if savings_rate < 10:
 archetypes.append(f"THE YOLO SPENDER (savings rate {savings_rate:.1f}%)")
if savings_rate > 40:
 archetypes.append(f"THE DISCIPLINED SAVER (savings rate {savings_rate:.1f}%)")
```

```python
# Bonus Archetype:
if (inv_pct > 15) and (food_pct > 25):
        archetypes.append("Indians are savers and hence are investors. But most importantly, we love food!")
```
